```python
import nibabel as nib
import numpy as np
from pathlib import Path
# Change path to 'labelsTr' to access the actual ground truth segmentation files
label_dir = Path(r"C:\DataSets\Medical_Segmentation_Decathlon\Task03_Liver\Task03_Liver\labelsTr")

for i, file_path in enumerate(label_dir.glob("*.nii.gz")):
    # Skip hidden files like ._liver_132.nii.gz
    if file_path.name.startswith("."):
        continue
        
    # Load the segmentation label volume
    label_img = nib.load(file_path)
    label = label_img.get_fdata()
    
    # Count voxels where the class is 2 (Tumor)
    tumor_voxels = (label == 2).sum()
    
    print(f"Val {i}: {file_path.name} | tumor voxels = {tumor_voxels}")
```
